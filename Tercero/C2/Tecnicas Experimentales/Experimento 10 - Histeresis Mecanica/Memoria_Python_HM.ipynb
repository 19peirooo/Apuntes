# **Librerias y Dependencias**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress 
import pandas as pd
```

# **Parte 1**

```python
g_func = lambda D,J,L: (D*L)/J
j_func = lambda R: (np.pi * R**4)/2
```

## *Datos*

```python
err_r = 0.001 #m
err_F = 0.05 #N
err_phi = np.deg2rad(2)
```

```python
angulos = np.array([np.deg2rad(n) for n in range(10,35,5)]) #º
```

```python
L_grande = 0.5 #m
L_pequeña = 0.3 #m
r = 0.1 #m
R = 0.001 #m
```

```python
fuerzas = {
    "Aluminio Grande": [0.35, 0.51, 0.75, 1.0, 1.2],
    "Laton": [0.1, 0.12, 0.21, 0.3, 0.4],
    "Cobre": [0.1, 0.2, 0.29, 0.4, 0.45],
    "Aluminio Pequeño": [0.1, 0.2, 0.3, 0.42, 0.59]
} #N
```

```python
for phi in angulos:
    print(f"φ = {phi:.5f} ± {err_phi:.5f}")
```

## _Varilla 1: Aluminio Grande_

```python
F = fuerzas["Aluminio Grande"]
M = np.array([f*r for f in F])
err_M = np.array([np.sqrt((r*err_F)**2 + (f*err_r)**2) for f in F])
```

```python
for f in F:
    print(f"F = {f} ± {err_F}")
```

```python
for i in range(len(M)):
    print(f"M = {M[i]:.2e} ± {err_M[i]:.5f}")
```

```python
m_regress,c_regress,r_sq,p_val,error_D = linregress(angulos,M)
plt.scatter(angulos,M, label="datos")
plt.plot(angulos, m_regress*angulos + c_regress, label="fit", color="g")
plt.xlabel("φ (rad)")
plt.ylabel("M (Nm)")
plt.legend()
plt.show()
```

```python
D = m_regress
D_aluminio = D # Lo necesito para parte 3
err_D_aluminio = error_D
J = j_func(R)
G = g_func(D,J,L_grande)
```

```python
err_J = J * 4 * (err_r/R)
err_G = G * np.sqrt((error_D/D)**2 + (err_r/L_grande)**2 + (err_J/J)**2)
```

```python
print(f"D = {D:e} ± {error_D:.5f} Nm/rad")
print(f"J = {J:e} ± {err_J:.4e} m^4")
print(f"G = {G:e} ± {err_G:.4e} Nm/rad")
```
