## _a) Graficar B(z)_

### Iman de Ferrita

```python
z = iman_ferrita["z (m)"]
B = iman_ferrita["B (T)"]

plt.plot(z, B, marker='o')
plt.xlabel("z (m)")
plt.ylabel("B (T)")
plt.title("Campo magnético del imán de ferrita")
plt.grid()

plt.show()
```

### Iman de Neodimio

```python
z = iman_neodimio["z (m)"]
B = iman_neodimio["B (T)"]

plt.plot(z, B, marker='o')
plt.xlabel("z (m)")
plt.ylabel("B (T)")
plt.title("Campo magnético del imán de neodimio")
plt.grid()
plt.show()
```

## b) Comparacion con valores teoricos

```python
B_iman = lambda M,V,R,z: (mag_perm/(4*np.pi))*(2*M*V/(z**2 + R**2)**1.5)
```

### Iman de Ferrita

```python
z = iman_ferrita["z (m)"]
B_experimental = iman_ferrita["B (T)"]
B_ferrita = [B_iman(M_ferrita,V_ferrita,R_ferrita,z_val) for z_val in z]

iman_ferrita["B_Teorico (T)"] = B_ferrita

df_ferrita = pd.DataFrame(iman_ferrita)
df_ferrita
```

```python
for b in B_ferrita:
    print(f"{b*1000:.2f}")
```

```python
plt.plot(z, B_experimental, marker='o', label = "Medido")
plt.plot(z, B_ferrita, marker="x", label="Calculado")
plt.xlabel("z (m)")
plt.ylabel("B (T)")
plt.title("Campo magnético del imán de ferrita")
plt.grid()
plt.legend()
plt.show()
```

### Iman de Neodimio

```python
z = iman_neodimio["z (m)"]
B_experimental = iman_neodimio["B (T)"]
B_neodimio = [B_iman(M_neodimio,V_neodimio,R_neodimio,z_val) for z_val in z]

iman_neodimio["B_Teorico (T)"] = B_neodimio

df_neodimio = pd.DataFrame(iman_neodimio)
df_neodimio
```

```python
plt.plot(z, B_experimental, marker='o', label = "Medido")
plt.plot(z, B_neodimio, marker="x", label="Calculado")
plt.xlabel("z (m)")
plt.ylabel("B (T)")
plt.title("Campo magnético del imán de neodimio")
plt.grid()
plt.legend()
plt.show()
```
