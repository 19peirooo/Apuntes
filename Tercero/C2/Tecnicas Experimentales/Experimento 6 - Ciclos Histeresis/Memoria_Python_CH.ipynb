# _Librerias y Dependencias_

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# **DATOS**

```python
cm3_160Hz = pd.read_csv("data/5cm_160Hz/5cm_160Hz_01.csv", sep=';', decimal=',', skiprows=[1])

cm5_160Hz = pd.read_csv("data/3cm_160Hz/3cm_160Hz_02.csv", sep=';', decimal=',', skiprows=[1])

cm5_80Hz = pd.read_csv("data/5cm_80Hz_01.csv", sep=';', decimal=',', skiprows=[1])

iman1 = pd.read_csv("data/iman_1/histeresis_iman_1.2_01.csv", sep=';', decimal=',', skiprows=[1])

iman2 = pd.read_csv("data/iman_2/histéresis_iman_2_01.csv", sep=';', decimal=',', skiprows=[1])
```

# **Ciclo Histeresis 3cm 160Hz**

```python
A = cm3_160Hz['average(A)']
B = cm3_160Hz['average(B)']

A_c = A + 0.01
```

```python
plt.figure()

plt.plot(A_c, B)

plt.xlabel("Voltaje A (V)")
plt.ylabel("Voltaje B (V)")
plt.title("Ciclo Histeresis 3cm a 160Hz")
plt.grid(True)

plt.axhline(0, color="black")  # eje horizontal
plt.axvline(0, color="black")  # eje vertical
plt.xlim(-0.04, 0.04)  # ajusta este rango

plt.tight_layout()
plt.show()
```
