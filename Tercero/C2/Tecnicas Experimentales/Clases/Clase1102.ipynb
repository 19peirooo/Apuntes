```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## Datos

```python
n = 5
M = np.array([100,80,60,40,20]).reshape((-1,1))
F = np.array([0.52,0.4,0.3,0.21,0.12])
```

```python
plt.scatter(M,F, label="data")
```

## Plottear Lineas de Regresion

```python
model = LinearRegression().fit(M,F)
r_sq = model.score(M,F)
m = model.coef_
c = model.intercept_
print(f"m: {m}. c: {c}. R^2: {r_sq}")

plt.scatter(M,F, label="data")
plt.plot(M, m*M + c, label="fit", color="g")
plt.legend()
plt.show()
```

## Calculos de E y D

```python
sum_mF = 0
sum_m2 = 0
mean_m = np.mean(M)
mean_F = np.mean(F)

for i in range(n):
    sum_mF += M[i]*F[i]
    sum_m2 += M[i]**2

E = sum_mF - (n*mean_m*mean_F)
D = sum_m2 - (n*mean_m**2)

print(f"Media m: {mean_m}. Media F: {mean_F}. E: {E}. D:{D}")
```

# Resultados

```python
print(f"m: {m}")
print(f"c: {c}")
print(f"R^2: {r_sq}")
print(f"r: {r_sq**0.5}")
print(f"E: {E}")
print(f"D: {D}")
```
