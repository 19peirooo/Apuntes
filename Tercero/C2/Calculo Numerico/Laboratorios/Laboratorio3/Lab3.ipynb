```python
def punto_fijo_r(g, x0, tol=1.0e-6, maxit=6, verbose=True):
    xk = x0
    errores = []
    r = None

    if verbose:
        print("k\t\tx_k\t\tcota error\t\tr")
        print(f"0\t\t{xk:.8f}\t")

    for k in range(1, maxit):
        xprev = xk
        xk = g(xprev)

        error = abs(xk - xprev)
        errores.append(error)

        if len(errores) >= 3:
            e3 = errores[-1]
            e2 = errores[-2]
            e1 = errores[-3]

            r = np.log(e2/e3)/np.log(e1/e2)

        if verbose:
            print(f"{k}\t\t{xk:.8f}\t{error:e}\t{r}")

        if error < tol:
            break
    else:
        xk = None
        print(f"Número máximo de iteraciones {maxit} alcanzado")

    return xk
```

### Intevalo [c, infinito]

Decrece monotonamente al atractor c en forma de escalera

```python
x0 = 3

punto_fijo_r(g,x0,verbose=True)
```

### Intevalo [b, c] con x0 = 1

Crece monotonamente al atractor c en forma de escalera

```python
x0 = 1

punto_fijo_r(g,x0,verbose=True)
```

### Intevalo [a, b] con x0 = 0.15

Decrece monotonamente al atractor a en forma de espiral

```python
x0 = 0.15

punto_fijo_r(g,x0,verbose=True)
```
