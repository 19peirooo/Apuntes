```python
import numpy as np
import matplotlib.pyplot as plt
```

## a)

```python
def calc_5p(f,x,h):

    calc1 = f(x - (2*h))
    calc2 = 8 * f(x - h)
    calc3 = 8 * f(x + h)
    calc4 = f(x + (2*h))

    return (calc1 - calc2 + calc3 - calc4)/(12*h)
```

## b) 

```python
f = lambda x: x*np.sin(x)
x0 = np.pi/2
true_val = 1

h_vals = [10**-n for n in range(1,4)]
errores = []
```

```python
print("h\t\tdf\t\terror")
for h in h_vals:

    df = calc_5p(f,x0,h)
    error = np.abs(true_val - df)
    errores.append(error)

    print(f"{h}\t\t{df}\t\t{error}")

e3 = errores[-1]
e2 = errores[-2]
e1 = errores[-3]

r = np.log(e2/e3)/np.log(e1/e2)

print(f"r = {r}")
```

El orden de convergencia es $O(h)$

## c)

```python
h_vals = [10**-n for n in range(1,7)]
errores2 = []

for h in h_vals:

    df = calc_5p(f,x0,h)
    error = np.abs(true_val - df)
    errores2.append(error)

    print(f"{h}\t\t{df}\t\t{error}")
```

```python
plt.loglog(h_vals, errores2)

plt.xlabel("h")
plt.ylabel("error")
plt.gca().invert_xaxis()

plt.show()
```

Para h mas pequeñas, el error empieza a incrementar de nuevo. Esto ocurre porque una vez pasado el h optimo, domina el error de redondeo sobre el error de discretización por lo que el metodo comienza a fallar
