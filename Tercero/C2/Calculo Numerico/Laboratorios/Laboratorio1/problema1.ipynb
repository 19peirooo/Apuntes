## _a) Evaluar f(x)_

$f(x) = \frac{1}{\sqrt{x^2 +1} - x}$

```python
fcc = lambda x: 1/(np.sqrt(x**2 + 1) - x) # Ocurre cancelación catastrofica

def evaluate_function(f):

    x_vals = [10**n for n in range(4,11)]

    for x_val in x_vals:
        try: 
            print(f"x = {x_val:.0e} --> f(x) = {f(x_val)}")
        except Exception:
            print("ERROR DE DIVISION ENTRE ZERO")

evaluate_function(fcc)
```

Ocurre cancelación catastrofica cuando x**2 es muy grande, la maquina no la suficiente precision para detectar el + 1 asi que no va a tener efecto. Entonces en el denominador tenemos que:  

$ \sqrt{x^2} - x = 0$

Los Resultados dan infinito porque cuando te vas acercando a 0 y divides por un numero muy pequeño asi que el resultado es muy grande. Sin embargo, en $10^10$ nos da un error de division entre 0 porque ha ocurrido esta cancelación catastrofica ya que el denominador es 0

## _Modificar Algebraicamente f(x)_

Para resolver esto multiplicamos y dividimos por el conjugado:

$f(x) = \frac{1}{\sqrt{x^2 +1} - x} = \frac{1}{\sqrt{x^2 +1} - x}·\frac{\sqrt{x^2 +1} + x}{\sqrt{x^2 +1} + x} = \frac{\sqrt{x^2 +1} + x}{1} = \sqrt{x^2 +1} + x$ 

```python
f = lambda x: np.sqrt(x**2 + 1) + x

def evaluate_function(f):

    x_vals = [10**n for n in range(4,10)]

    for x_val in x_vals:
        print(f"x = {x_val:e} --> f(x) = {f(x_val)}")

evaluate_function(f)
```

Calculamos Errores Relativos y Absolutos

```python
def evaluate_function_with_errors(f,fcc):

    x_vals = [10**n for n in range(4,11)]

    for x_val in x_vals:
        try:
            f_x = f(x_val) 
            f_x_cc = fcc(x_val)
            err_abs = np.abs(f_x - f_x_cc)
            err_rel = (err_abs/f_x)*100
            print(f"x = {x_val:.0e}\tf(x) = {f_x}\tfcc(x)={f_x_cc}\tError Absoluto = {err_abs:.2e}\tError Relativo = {err_rel:.2e}%")
        except Exception:
            print("ERROR DE DIVISION ENTRE ZERO")

evaluate_function_with_errors(f,fcc)
```

Nota: Errores infinitos porque cuando sumas o restas con infinito pues da infinito
