```python
import numpy as np
```

## _a) $x^2 - y^2$ con enteros_

```python
x_int = 1234567891234567
y_int = 1234567891234566

res_int = x_int**2 - y_int**2

print(f"x={x_int}, y = {y_int} --> x^2 - y^2 = {res_int}")
```

## _b) $x^2 - y^2$ con floats_

```python
x_float = 1234567891234567.
y_float = 1234567891234566.

res_float = x_float**2 - y_float**2

print(f"x={x_float}, y = {y_float} --> x^2 - y^2 = {res_float}")
```

No da el mismo resultado por que los enteros que con los numeros que coma flotantes porque python usa 32 bits para la representacion de numero en coma flotante asi que ocurre un overflow ya que nuestros numeros son mas grandes

```python
err_abs = np.abs(res_float - res_int)
err_rel = (err_abs/res_int)*100

print(f"Error Absoluto: {err_abs} --> Error Relativo: {err_rel:.3f}%")
```

## _c) Resolucion Con Identidad notable_

```python
res_notable = (x_float - y_float) * (x_float + y_float)

print(f"x={x_float}, y = {y_float} --> x^2 - y^2 = {res_notable}")
```

No da el mismo resultado que en el aparato b). Esto ocurre que cuando trabajamos con numeros en coma flotante. Puede ocurrir una falta de asociatividad. Esto se refiere que el orden en el que se hacen los calculos si afecta al resultado
