# Pregunta 1 A

```python
f = lambda x: x**3 - 2*x - 5
df = lambda x: 3*x**2 - 2
x0 = 0
```

```python
def newton(f,df,x0,tol=1e-6,maxit=100, verbose=True):

    k = 0
    xk = x0
    xk1 = None
    error = None

    if verbose:
        print(f"k\t\txk\t\txk1\t\terr")

    while (k < maxit):

        fxk = f(xk)
        dfxk = df(xk)

        if fxk != 0:

            xk1 = xk - (fxk/dfxk)

            error = np.abs(xk1 - xk)
            if verbose: 
                print(f"{k}\t\t{xk:e}\t\t{xk1:e}\t\t{error:e}")

            if error < tol:
                break

        xk = xk1
        k += 1

    else:
        print("Numero Maximo de Iteraciones")

    return xk1
```

```python
root = newton(f,df,x0)
root
```

# Problema 1 B

```python
x1 = 3
```

```python
def newton_secante(f,df,x0,x1,tol=1e-6,maxit=100, verbose=True):

    k = 0
    x_prev = x0
    x = x1
    x_next = None
    error = None

    if verbose:
        print(f"k\t\txk-1\t\txk\t\txk+1\t\terr")

    while (k < maxit):

        f_x = f(x)
        df_x = df(x)

        if f_x != 0:

            if df_x < 1e-5:
                f_xprev = f(x_prev)
                x_next = x - f_x*((x - x_prev)/(f_x - f_xprev))
            else:
                x_next = x - (f_x/df_x)

            error = np.abs(x_next - x)
            if verbose:
                print(f"{k}\t\t{x_prev:e}\t\t{x:e}\t\t{x_next:e}\t\t{error:e}")

            if error < tol:
                break
        
        x_prev = x
        x = x_next
        k += 1

    else:
        print("Numero Maximo de Iteraciones")

    return x_next
```

# Problema 1C

```python
dcf = lambda f,x,h: (f(x+h) - f(x-h))/2*h
```

```python
def newton_dcf(f,dcf,x0,h,tol=1e-6,maxit=100, verbose=True):

    k = 0
    xk = x0
    xk1 = None
    error = None

    if verbose:
        print(f"k\t\txk\t\txk1\t\terr")

    while (k < maxit):

        fxk = f(xk)
        dfxk = dcf(f,xk,h)

        if fxk != 0:

            xk1 = xk - (fxk/dfxk)

            error = np.abs(xk1 - xk)
            if verbose: 
                print(f"{k}\t\t{xk:e}\t\t{xk1:e}\t\t{error:e}")

            if error < tol:
                break

        xk = xk1
        k += 1

    else:
        print("Numero Maximo de Iteraciones")

    return xk1
```

```python
def richardson(Dh,Dh2,p):
    return (2**p * (Dh2 - Dh))/(2**p -1)
```

```python
def richardson_forward(f,x,h):
    Dh = forward(f,x,h)
    Dh2 = forward(f,x,h/2)
    p = 1

    return richardson(Dh,Dh2,p)
```
