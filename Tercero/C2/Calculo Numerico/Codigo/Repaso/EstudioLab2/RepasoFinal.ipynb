# **Iteración de Punto Fijo**

```python
f = lambda x: np.cos(x) - x
g = lambda x: np.cos(x)
x0 = 1
```

```python
def punto_fijo(g, x0, tol=1e-6, maxit=100, verbose=True):

    xprev = x0
    x = None
    error = None
    k = 0

    if verbose:
        print("k\t\tx\t\terror")

    while k < maxit:

        x = g(xprev)

        error = np.abs(x - xprev)

        if verbose:
            print(f"{k}\t\t{x}\t\t{error}")

        if error < tol:
            break

        xprev = x
        k += 1

    else:
        print("Maxit reached")

    return x
```

```python
def punto_fijo_errores(g, x0, tol=1e-6, maxit=100, verbose=True):

    xprev = x0
    x = None
    error = None
    r = None
    C = None
    k = 0
    errores = []

    if verbose:
        print("k\t\tx\t\terror\t\tr\t\tC")

    while k < maxit:

        x = g(xprev)

        error = np.abs(x - xprev)
        errores.append(error)

        if len(errores) >= 3:
            e3 = errores[-1]
            e2 = errores[-2]
            e1 = errores[-3]

            r = np.log(e2/e3)/np.log(e1/e2)
            C = e3 / (e2)**r

        if verbose:
            print(f"{k}\t\t{x}\t\t{error}\t\t{r}\t\t{C}")

        if error < tol:
            break

        xprev = x
        k += 1

    else:
        print("Maxit reached")

    return x
```

```python
root = punto_fijo_errores(g,x0)
root
```

# **Newton**

```python
f = lambda x: np.cos(x) - x
df = lambda x: -np.sin(x) - 1
x0 = 1
```

```python
def newton(f,df,x0,tol=1e-6,maxit=100, verbose=True):

    xprev = x0
    x = x0
    error = None
    k = 0

    if verbose:
        print("k\t\tx\t\terror")

    while k < maxit:

        f_x = f(xprev)
        df_x = df(xprev)

        if (df_x != 0):

            x = xprev - (f_x/df_x)

            error = np.abs(x - xprev)

            if verbose:
                print(f"{k}\t\t{x}\t\t{error}")

            if error < tol:
                break

        k += 1
        xprev = x

    else:
        print("Maxit reached")

    return x
```

```python
def newton_errores(f,df,x0,tol=1e-6,maxit=100, verbose=True):

    xprev = x0
    x = x0
    error = None
    r = None
    C = None
    k = 0
    errores=[]

    if verbose:
        print("k\t\tx\t\terror\t\tr\t\tC")

    while k < maxit:

        f_x = f(xprev)
        df_x = df(xprev)

        if (df_x != 0):

            x = xprev - (f_x/df_x)

            error = np.abs(x - xprev)

            errores.append(error)

            if len(errores) >= 3:
                e3 = errores[-1]
                e2 = errores[-2]
                e1 = errores[-3]

                r = np.log(e2/e3)/np.log(e1/e2)
                C = e3 / (e2)**r

            if verbose:
                print(f"{k}\t\t{x}\t\t{error}\t\t{r}\t\t{C}")

            if error < tol:
                break

        k += 1
        xprev = x

    else:
        print("Maxit reached")

    return x
```

```python
root = newton_errores(f,df,x0)

root
```

# **Secante**

```python
def secante(f,x1,x2,tol=1e-6,maxit=100,verbose=True):
    
    k=0
    xprev = x1
    x = x2
    xnext = None
    error = None

    if verbose:
        print("k\t\txnext\t\terror")

    while k < maxit:

        f_xprev = f(xprev)
        f_x = f(x)

        xnext = x - f_x * ((x - xprev)/(f_x - f_xprev))

        error = np.abs(xnext - x)

        if verbose:
            print(f"{k}\t\t{xnext}\t\t{error}")

        if error < tol:
            break

        xprev = x
        x = xnext
        k  += 1
    
    return xnext
```

```python
def secante_errores(f,x1,x2,tol=1e-6,maxit=100,verbose=True):
    
    k=0
    xprev = x1
    x = x2
    xnext = None
    error = None
    errores = []
    r = None
    C = None

    if verbose:
        print("k\t\txnext\t\terror\t\tr\t\tC")

    while k < maxit:

        f_xprev = f(xprev)
        f_x = f(x)

        xnext = x - f_x * ((x - xprev)/(f_x - f_xprev))

        error = np.abs(xnext - x)

        errores.append(error)

        if len(errores) >= 3:
            e3 = errores[-1]
            e2 = errores[-2]
            e1 = errores[-3]

            r = np.log(e2/e3)/np.log(e1/e2)
            C = e3 / (e2)**r

        if verbose:
            print(f"{k}\t\t{xnext}\t\t{error}\t\t{r}\t\t{C}")

        if error < tol:
            break

        xprev = x
        x = xnext
        k  += 1
    
    return xnext
```

```python
f = lambda x: np.log(x) + x
x1 = 1
x2 = 2
root = secante_errores(f,x1,x2)

root
```

# **Derivadas**

```python
def forward(f,x,h):
    return (f(x + h) - f(x))/h

def backward(f,x,h):
    return (f(x) - f(x - h))/h

def centrada(f,x,h):
    return f(x + h) - f(x - h)/(2*h)
```

```python
def richardson(Dh,Dh2,p):
    return ((2**p * Dh2) - Dh)/(2**p - 1)
```

```python
def richardson_forward(f,x,h):
    Dh = forward(f,x,h)
    Dh2 = forward(f,x,h/2)
    p = 1

    return richardson(Dh,Dh2,p)

def richardson_backward(f,x,h):
    Dh = backward(f,x,h)
    Dh2 = backward(f,x,h/2)
    p = 1

    return richardson(Dh,Dh2,p)

def richardson_centrada(f,x,h):
    Dh = centrada(f,x,h)
    Dh2 = centrada(f,x,h/2)
    p = 2

    return richardson(Dh,Dh2,p)
```
