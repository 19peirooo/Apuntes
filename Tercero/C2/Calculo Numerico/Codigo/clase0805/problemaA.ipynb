## **Numericamente**

$
\int_a^b f(x)\,dx \approx
\frac{h}{2}
\left[
f(x_0)+2\sum_{i=1}^{n-1}f(x_i)+f(x_n)
\right]
$

donde

$
h=\frac{b-a}{n}
$

```python
def trapecio(y,x):

    n = len(x) - 1

    s = 0.
    
    for i in range(n):
        h_i = x[i+1] - x[i]
        s += 0.5 * h_i * (y[i] + y[i+1])

    return s
```

```python
def trapecio_opt(y,x):

    h = np.diff(x)
    
    return 0.5 * np.sum(h * (y[:-1] + y[1:]))
```

```python
def trapecio_equi(y, x):

    h = x[1]-x[0]
    n = len(x) - 1

    s = 0.
    
    for i in range(n):
        s += 0.5 * h * (y[i] + y[i+1])

    return s
```

```python
def trapecio_equi_opt(y,x):

    h = x[1] - x[0]
    
    area = 0.5 * h * (y[0] + (2.*np.sum(y[1:-1])) + y[-1])

    return area
```

```python
W = trapecio(y,x)
W_opt = trapecio_opt(y,x)
W_equi = trapecio_equi(y,x)
W_equi_opt = trapecio_equi_opt(y,x)

print(f"W = {W}J")
print(f"W_opt = {W_opt}J")
print(f"W_equi = {W_equi}J")
print(f"W_equi_opt = {W_equi_opt}J")
```
