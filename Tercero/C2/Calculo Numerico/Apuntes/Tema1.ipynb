## <center>**Errores**</center>

La aproximación inherementemente produce errores

**Error Inevitable:** Aritmetica de Precisión Finita. Representación de reales

**Error de Redondeo:** Limitaciones de la aritmetica a redondear números reales

**Error de Truncamiento:** Al reemplazar un problema continuo por una aproximación discreta

### _Aritmetica de Precisión Finita_

Los ordenadores no pueden representar numeros con digitos infinitos. Se observa con el siguente ejemplo:

```python
print(1.1 + 0.1 == 0.2)
print(1.1 + 0.1)
```

Al evaluar 1.1 + 0.1 = 1.2 nos sale falso.

Llegamos a la conclusion de que solo nos podemos fiar de los 15 primeros decimales

### _Error Absoluto y Relativo_

Criterio:  
$x$ = Valor exacto  
$\tilde{x}$ = Valor aproximado

**Error Absoluto:** Mide cuanto nos desviamos

$\Delta x = \tilde{x} - x$

**Cota:** $\tilde{x} - \epsilon \leq x \leq \tilde{x} + \epsilon$

**Error Relativo:** Mide cuanto nos desviamos en proporcion a x

Si $x \neq 0$: $\delta x = \frac{|\Delta x|}{x}$

### _Propagación de Errores_

**Suma y Resta:** $|\Delta x| = |\Delta a | \pm |\Delta b|$

**Multiplicación y división:** $|\delta x| \geq |\delta a | + |\delta b|$

**Cancelación Catastrofica**: Cuidado si $|a - b| \approx 0 \rightarrow $ Error relativo grande

## <center>**Fenomenos de la aritmentica de Precisión Finita**</center>

### _Cancelación Catastrofica_

Para entender la cancelación catastrofica, usaremos el siguiente ejemplo:

$\lim_{x \to 0} \frac{1 - cos(x)}{x^2}$

El problema que tenemos es que la función no esta definida en 0. Podemos calcular la solución de forma analitica y de forma numerica.

**Solución Analitica**

```python
x = sympy.symbols('x')
f_sym = (1 - sympy.cos(x)) / x**2
limit_sym = sympy.limit(f_sym, x, 0)
print(limit_sym)
```

**Solución Numerica**

```python
f = lambda x: (1 - np.cos(x))/x**2
xi = (1.e-5, 1.e-6, 1.e-7, 1.e-8, 1.e-9, 1.e-10, 1.e-11)
for x in xi:
    print(f'f({x}) = {f(x)}')
```

Como se puede apreciar, a valores muy pequeños de x, la función tiende a 0 y no ha 0.5. La diferencia entre 1 - cos(x) es un numero muy cercano a 0 asi que por eso los valores tienden a cero.

La solución a esto es reformular la función. Un metodo muy común es multiplicar y dividir por el conjugado.

$\frac{1-cos(x)}{x^2} = \frac{1-cos(x)}{x^2} · \frac{1+cos(x)}{1+cos(x)} =  \frac{sin^2(x)}{(1+cos(x))x^2}$

```python
f = lambda x: (np.sin(x)**2)/(x**2 * (1 + np.cos(x)))
xi = (1.e-5, 1.e-6, 1.e-7, 1.e-8, 1.e-9, 1.e-10, 1.e-11)
for x in xi:
    print(f'f({x}) = {f(x)}')
```

### _Falta de Asociatividad_

**Falta de Asociatividad:** El orden en el que se realizan la operaciones puede afectar el resultado final

```python
x = 1.0e20
y = -1.0e20
z = 1.0
print(f'(x + y) + z = {(x + y) + z}')
print(f'x + (y + z) = {x + (y + z)}')
```

### _Precisión de la Maquina_

Existe la no unicidad del cero

Existen numeros distintos de cero

**Epsilón de la Maquina:** Numero positivo mas pequeño tal que $1 + EPS \neq 1$

Se calcula de la siguiente forma:

```python
x = 1.0
while (1.0 + x != 1.0):
    x = x / 2.0
EPS = 2.0 * x
print(f'EPS = {EPS}')
print(f'x = {x}')
```

## <center>**Notación Cientifica Normalizada**</center>

**Notación Cientifica:** Es la forma de representar numeros reales en sistemas informáticos  
Garantiza que el primer digito significativo este en una posición no nula

**Normalizada:** El primer numero significativo esta a la izquierda del punto

$x = \pm r · b^n$  

$\pm$: Signo  
b: Base del sistema  
r: Mantisa ($0 \leq r \leq b$)  
n: Exponente, un entero

Para representar exactamente un numero real, necesitariamos una mantisa y exponente infinitos. Como esto no es posible, se definen limites

**Limites de n**: $L \leq n \leq U$

**Overflow:** Ocurre cuando el exponente excede cierto limite superior  
**Underflow:** Ocurre cuando el exponente es inferior al límite inferior

### _Binario_

Los sistema informaticos usan el sistema binario para almacenar numeros en binario. Existen 2 formas de representar numeros en coma flotante

**Precisión Simple (32 bits):** 1 bit para el signo(bit 1), 8 bits para el exponente(bits 2-9) y 23 para la mantisa (bits 10-32)   
**Precisión Doble (64 bits):** 1 bit para el signo(bit 1), 11 bits para el exponente(bits 2-12) y 52 para la mantisa (bits 13-64)
