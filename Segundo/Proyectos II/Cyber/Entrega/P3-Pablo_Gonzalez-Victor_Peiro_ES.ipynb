### Convert problem to binary: normal vs attack (OPTIONAL)

```python
awid_train['class'] = awid_train['class'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
awid_test['class'] = awid_test['class'].apply(lambda x: 'normal' if x == 'normal' else 'attack')
```

### Transform categorical features to ordinal values so most ml algorithms can be applied


```python
from sklearn.preprocessing import OrdinalEncoder

# Columns with str dtype are categorical.
categorical_columns = [column for column, column_type in dtypes.items() if column_type is str]

# Missing values are encoded as -1. Not known values during test transformation are also encoded as -1
encoder = OrdinalEncoder(encoded_missing_value=-1, handle_unknown='use_encoded_value', unknown_value=-1)

# Apply Ordinal Encoder
awid_train[categorical_columns] = encoder.fit_transform(awid_train[categorical_columns])
awid_test[categorical_columns] = encoder.transform(awid_test[categorical_columns])
```

```python
X_awid_train = awid_train.iloc[:,:-1]
y_awid_train = awid_train['class']
X_awid_test = awid_test.iloc[:,:-1]
y_awid_test = awid_test['class']
```

## Preprocesamiento de Datos

Primero eliminamos las columnas cuya varianza sea 0

```python
#Eliminacion de columnas cuya varianza sea 0
varianzas = np.var(X_awid_train, axis= 0)
columnas_eliminadas = varianzas[varianzas==0].index
X_awid_train = X_awid_train.drop(columns= columnas_eliminadas)
X_awid_test = X_awid_test.drop(columns= columnas_eliminadas)
```

```python
columnas_eliminadas
```

Despues eliminamos las columnas que sean todo nulo

```python
#Eliminacion de columnas cuya columna entera sea nula
columnas_null = X_awid_train.columns[X_awid_train.isnull().all()]
X_awid_train = X_awid_train.drop(columns= columnas_null)
X_awid_test = X_awid_test.drop(columns= columnas_null)
```

```python
columnas_null
```

Se rellenan valores nulos. -1 en columnas categoricas y la media en las columnas numericas

```python
#Relleno los valores nulos con -1
categorical_columns = [col for col in categorical_columns if col in X_awid_train]
numerical_columns = X_awid_train.select_dtypes(include=[np.number]).columns
mean = X_awid_train[numerical_columns].mean()

X_awid_train[categorical_columns] = X_awid_train[categorical_columns].fillna(-1)
X_awid_test[categorical_columns] = X_awid_test[categorical_columns].fillna(-1)

X_awid_train[numerical_columns] = X_awid_train[numerical_columns].fillna(mean)
X_awid_test[numerical_columns] = X_awid_test[numerical_columns].fillna(mean)
```

## Normalizacion de los datos

```python
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

X_awid_train_norm = pd.DataFrame(scaler.fit_transform(X_awid_train),columns=X_awid_train.columns, index= X_awid_train.index)
X_awid_test_norm = pd.DataFrame(scaler.transform(X_awid_test),columns=X_awid_test.columns, index= X_awid_test.index)
```

## Prueba 1 - KNN

kNN es un metodo basado en la simitud entre datos. Para ellos tenemos que seleccion un valor de k que sera el numero de puntos cercanos que se consideran para hacer la prediccion. Para ello se mide la distancia entre los datos y se identifican los que tiene mas cercanos.

Despues, se asigna una clase entre los k vecinos

```python
#TODO
# Apliquemos Knn
from sklearn.neighbors import KNeighborsClassifier

# Definir el modelo de clasificacion K-nn
neigh = KNeighborsClassifier(3)

# Entrenarlo
neigh.fit(X_awid_train_norm,y_awid_train)

# Obtener las predicciones para el conjunto de datos de testeo
y_est = neigh.predict(X_awid_test_norm)

# Obtener el error cuadratico medio y el precision
MSE_tst = mean_squared_error(y_awid_test,y_est)
accuracy = balanced_accuracy_score(y_awid_test,y_est)

# Imprimimos el error y precision del modelo
print('MSE : ' + str(MSE_tst))
print('Accuracy: ' + str(accuracy))
```

Precision de 82.38%
kNN es muy lento y no es lo mas eficaz para volumen altos de datos. Por ellos pasaremos a utilizar metodos de ensemble.

Los métodos de ensemble combinan múltiples modelos de aprendizaje automático para obtener predicciones más precisas y robustas. La idea principal es que varios modelos juntos pueden reducir errores, mejorar la generalización y minimizar el sobreajuste. Hay 3 tipos de ensemble pero nosotros solo vamos a aplicar 2 de ellos que son:

**Bagging**:
- Entrena múltiples modelos independientes en subconjuntos aleatorios de los datos.
- Los modelos son del mismo tipo.
- La predicción final se obtiene por votación.
- Reduce la varianza, útil para modelos que sobreajustan.
- Ejemplos:
    - RandomForestClassifier
    - BaggingClassifier

**Boosting**:
- Los modelos se entrenan secuencialmente, cada uno intentando corregir los errores del anterior.
- Da mayor peso a las observaciones mal clasificadas.
- Reduce el sesgo y mejora la precisión, pero puede ser más propenso a sobreajuste.
- Ejemplos:
    - GradientBoostingClassifier
    - AdaBoostClassifier

### Prueba 3.1 - Bagging

```python
# Apliquemos Ensemble de Bagging con BaggingClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import mean_squared_error, balanced_accuracy_score
```

#### Datos no normalizados

El Bagging Classifier (Bootstrap Aggregating Classifier) es una técnica de aprendizaje de ensamblado (ensemble learning) que mejora la estabilidad y precisión de los modelos de machine learning al reducir la varianza. Se encuentra implementado en Scikit-learn dentro del módulo sklearn.ensemble

**Uso de sus hiperparámetros**
- estimator: especifíca el modelo base (decisionTreeClassifier)
- n_estimator: número de clasificadores base a entrenar.
- random_state: es clave en machine learning porque permite controlar la aleatoriedad en el entrenamiento de modelos. En el contexto de BaggingClassifier y otros algoritmos, define la semilla del generador de números aleatorios, asegurando que los resultados sean reproducibles.

```python
# Definir el modelo base
base_model = DecisionTreeClassifier(max_depth=1)

# Definir el modelo de Bagging
model = BaggingClassifier(estimator=base_model, n_estimators=100, random_state=42)

print("Training: ")
# Entrenar el modelo
model.fit(X_awid_train, y_awid_train)

# Obtener las predicciones para el conjunto de datos de testeo
y_est = model.predict(X_awid_test)

# Obtener el error cuadrático medio y la precisión
MSE_tst = mean_squared_error(y_awid_test, y_est)
accuracy = balanced_accuracy_score(y_awid_test, y_est)

# Imprimir el error y precisión del modelo
print('MSE : ' + str(MSE_tst))
print('Accuracy: ' + str(accuracy))
```

Se obtiene un 84.78% lo que no es un gran porcentaje de acierto. Esto se puede deber a que no elimina el sesgo del modelo base ya que es un algoritmo de bagging. Además, es computacionalmente costoso

## Prueba 3 - Ensembles de Boosting

### Prueba 3.1 - Ada Boosting

AdaBoost es un método de ensemble que combina múltiples clasificadores débiles ponderándolos de manera iterativa para mejorar la precisión global del modelo. 
Este funciona asignado los mismo pesos a todas las muestras de entrenamiento al principio. Despues de cada iteración, se le asigna mayor peso a aquellas muestras que fueron clasificadas incorrectamente. La precision final es la combinacion ponderada de todos los clasificadores.

Para ello hemos implementado un arbol de decisiones como el modelo base(clasificador débil). Este modelo funciona construyendo el arbol de decisiones donde cada nodo es una caracteristica es una caracteristica del conjuntos de datos.

Parametros de DecisionTreeClassifier que vamos a modificar:
- **max_depth**: Profundidad maxima que se recorrera del arbol de decisiones

Parametros de AdaBoostClassifier que vamos a modificar:
- **estimator**: El clasificador base que vamos a usar
- **n_estimators**: El número de clasificadores debiles que vamos a usar 
- **learning_rate**: El peso aplicado a cada clasificador cada iteración

### Prueba 3.2 - Gradient Boosting

GradientBoosting es otro metodo de ensemble que tambien esta basado en arboles de decision. Este en vez de ajustar pesos, intenta minimizar el error ajustando el modelo en funcion del gradiente de la función de perdida. Primero, se empieza con un modelo inicial simple y se calculan los residuos entre la prediccion y la realidad. Despues, se van construyendo nuevos modelos para predecir los residuos y nuevas predicciones sumando los nuevos modelos y ajustando una taza de aprendizaje

Ventajas:

- Funciona bien frente a datos complejos y grandes volumenes de datos

Desventajas:

- Mas lento que AdaBoost

Parametros a modificar:

#### Busco los mejores hiperparametros

Al principio pensamos en usar GridSearch para buscar los mejores parametros. Sin embargo nos dimos cuenta que necesitamos una solucion mas rapida ya que GridSearch a pesar de ser muy precisa, era muy lenta. Por ello pasamos a usar Optuna.

Para poder usar optuna, primero tenemos que definir el espacio de busqueda estableciendo los hiperparametros a modificar y sus rangos. Despues se van sugeriendo hiperparametros y entrenando modelos para ver cual da mayor precision.

Funciona por un sistema de trails en el cuando se puede ejecutar simultaneamente haciendo la busqueda mas eficiente

```python
import optuna
from sklearn.model_selection import cross_val_score
```

```python
def objective(trial):
    parametros = {
        "n_estimators": [50, 100],
        "learning_rate": [0.01, 0.1, 0.3, 0.5],
        "max_depth": [3, 5, 7]
    }
    
    n_estimators = trial.suggest_categorical("n_estimators", parametros["n_estimators"])
    learning_rate = trial.suggest_categorical("learning_rate", parametros["learning_rate"])
    max_depth = trial.suggest_categorical("max_depth", parametros["max_depth"])
    
    model = GradientBoostingClassifier(
        n_estimators=n_estimators,
        learning_rate=learning_rate,
        max_depth=max_depth,
        random_state=42
    )
    
    score = cross_val_score(model, X_awid_train_norm, y_awid_train, cv=5, scoring='balanced_accuracy').mean()
    print(f"Trial {trial.number}: Accuracy = {score}")
    return score

# Run Optuna optimization
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=20,n_jobs=-1)

# Best hyperparameters
best_params = study.best_params
print("Best Parameters:", best_params)
```

El mejor trail ha sido el trail 15. Que nos ha dado una precision de 98,46%. Hemos conseguido esto con los siguiente parametros:
- n_estimators = 50
- learning_rate = 0.5
- max_depth = 3
