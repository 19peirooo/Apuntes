# 2. Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0, 2, 100)

print(x[0:10])

plt.plot(x, x, label='recta')
plt.plot(x, x**2, label='parábola')
plt.plot(x, x**3, label='cúbica')

plt.xlabel('x')
plt.ylabel('y')

plt.title("Un plot sencillo")

plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# y = x^3 - x**5 entre -1 y 1

x = np.linspace(-1,1,60)
y = x**3 - x**5
plt.figure(figsize=(8,5)) # tamaño de la gráfica en pulgadas
plt.title('Gráfica de $y = x^3 - x^5$') # Título icluyendo una fórmula LaTeX
plt.xlabel('$x$') # título del eje horizontal
plt.ylabel('$y$') # título del eje  vertical
plt.plot(x, y, marker='.', label=r"$x^3-x^5$", color = "red") # leyenda
plt.legend() # muestra la leyenda
plt.show()   # muestra la gráfica
```

```python
# Suma de señales como en el ejemplo que vimos en la parte de introducción

def sumsignals(f1=7.0, f2=68.0):
    max_time = 0.5
    plt.figure(figsize=(8,5))
    times = np.linspace(0, max_time, 1000)
    signal = np.sin(2 * np.pi * f1 * times) + 0.15 * np.sin(2 * np.pi * f2 * times)
    with plt.style.context("ggplot"):
        plt.plot(signal, color="blue")
        plt.xlabel("Tiempo ($t$)")
        plt.title("Suma de señales")

sumsignals()
```

### Figuras múltiples y ejes

```python
def coseno(t):
    return np.cos(2*np.pi*t)

# Función cos multiplicada por una exponencial decreciente
def cosdec(t):
    return np.exp(-t) * np.cos(2*np.pi*t)

t1 = np.arange(0.0, 5.0, 0.1)   # Separación de los puntos en el eje X
t2 = np.arange(0.0, 5.0, 0.04)  # Escala para la línea

plt.figure(figsize=(12,4))
plt.figure(1)
plt.subplot(121)
plt.plot(t1, cosdec(t1), 'bo', t2, cosdec(t2), 'k')

plt.subplot(122)
plt.plot( t2, coseno(t2), 'r--')
plt.show()
```

### Histogramas. Añadir texto y grid

```python
mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)

# Histograma, el segundo parámetro es el número de bins
n, bins, patches = plt.hist(x, 25, density=1, facecolor='b', alpha=0.5)


plt.xlabel('CI')
plt.ylabel('Probabilidad')
plt.title('Distribución del CI')
plt.text(60, .02, r'$\mu=100,\ \sigma=15$')
plt.axis([40, 160, 0, 0.03])
plt.grid(True)
plt.show()
```

```python
np.random.seed(2022)

# example data
mu = 100  # mean of distribution
sigma = 15  # standard deviation of distribution
x = mu + sigma * np.random.randn(437)

num_bins = 50

fig, ax = plt.subplots()

# the histogram of the data
n, bins, patches = ax.hist(x, num_bins, density=1)

# add a 'best fit' line applying the Gaussian formula 
y = ((1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-0.5 * ((bins - mu)/ sigma )**2))

ax.plot(bins, y, '--r')
ax.set_xlabel('Smarts')
ax.set_ylabel('Probability density')
ax.set_title(r'Histogram of IQ: $\mu=100$, $\sigma=15$')

# Tweak spacing to prevent clipping of ylabel
fig.tight_layout()
plt.show()
```

### Diferentes tipos de ejes a demás del lineal

```python
from matplotlib.ticker import NullFormatter  # useful for `logit` scale

# Fixing random state for reproducibility
np.random.seed(2022)

# make up some data in the interval ]0, 1[
y = np.random.normal(loc=0.5, scale=0.4, size=1000)  # np.random.normal(mean, std, size)
y = y[(y > 0) & (y < 1)]
y.sort()
x = np.arange(len(y)) # to generate the X axis

# plot with various axes scales

plt.figure(figsize=(8,3))

# linear
plt.subplot(121)
plt.plot(x, y)
plt.yscale('linear')
plt.title('linear')
plt.grid(True)


# log
plt.subplot(122)
plt.plot(x, y)

plt.title('log')
plt.grid(True)


# Format the minor tick labels of the y-axis to avoid too many labels.
plt.gca().yaxis.set_minor_formatter(NullFormatter())
# Adjust the subplot layout, because the logit one may take more space
# than usual, due to y-tick labels like "1 - 10^{-3}"

plt.subplots_adjust(top=1.2, bottom=0.08, left=0.10, right=0.95, hspace=0.25, wspace=0.35)

plt.show()
```

Pintar la función:  𝑦=0.25⋅(𝑥+4)⋅(𝑥+2)⋅(𝑥−3)
 
Vamos a restringir la definición de la función para valores de x entre -5 y +5 y pintaremos 1000 puntos que matplotlib unirá con una línea.

```python
#Solución
import numpy as np
import matplotlib.pyplot as plt  

x = np.linspace(-5, 5, 1000)

y = 0.25 *(x+4)* (x+2) * (x-3)

#Print function
plt.plot(x,y, color="navy", linewidth=3, label="función")


#Print green axis
x2= np.linspace(-8, 8, 1000)
y2=x2*0
plt.plot (x2,y2,color="lime", linestyle ='--', label="eje horizontal")


y3= np.linspace(-20, 35, 1000)
x3= y3*0
plt.plot (x3,y3,color="darkgreen",linestyle='--',label="eje vertical")

#print roots
lista = [-4,-2,3]
rootsx=np.array(lista)
rootsy = rootsx *0
plt.plot (rootsx, rootsy, 'r*', markersize=10, label="raíces")
#con lista también funciona: plt.plot (lista, rootsy, 'r*', markersize=10)

#Title
plt.title("Función de tercer grado")

plt.legend()

plt.text(-5,3 ,"R1=-4")
plt.text(-2,3 ,"R2=-2")
plt.text(1.5,3 ,"R3=3")


#Required to put 6 to obtain a list from -5 to 5
listx = range(-5,6)
myticks=np.array(listx)
#alternative: np.arange(-5,6)
#set x axis values
#plt.xticks(listx)
plt.xticks(myticks)

plt.show()
```
