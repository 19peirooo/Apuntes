```python
import numpy as np
import matplotlib.pyplot as plt
import math
%matplotlib inline
```

```python
numPuntos = 50

x = np.linspace(-4,4,numPuntos)
y = np.exp(x)
plt.plot(x,y)
plt.show()
```

```python
a = 1
def taylor_series(x,a,nterm):
    serie = np.zeros(len(x))
    for i in range(nterm):
        serie += (np.exp(a)/math.factorial(i))*((x-a)**i)
    return serie


y_taylor_chapuza = np.exp(a)+(np.exp(a)*(x-a))+(np.exp(a))*((x-a)**2)/(math.factorial(2))
y_taylor = taylor_series(x,a,9)
plt.plot(x,y, label = "f(x)")
plt.plot(x,y_taylor_chapuza, label="Aproximacion Chapuza")
plt.plot(x,y_taylor,label = "Aproximacion")
plt.legend()
plt.show()
```

```python
def aprox_e(n):
    aprox = 0
    for i in range(n):
        aprox += 1/math.factorial(i)
    return aprox
n = 100

ap_e = aprox_e(100)
print(np.exp(1)-ap_e)

ap_e_lim = (1+(1/n))**n
print(np.exp(1)-ap_e_lim)
```
