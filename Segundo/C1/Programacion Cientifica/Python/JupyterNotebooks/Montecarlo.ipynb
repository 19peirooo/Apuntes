# Volumenes Montecarlo

```python
import numpy as np
import matplotlib.pyplot as plt
import math
import random
```

```python
def errorMC(areaExacta,areaCalculada):
    return abs(100 * (areaExacta - areaCalculada) / areaExacta)
```

## Para Calcular Volumen Cilindro

```python
def volCilindroMC(L,h,n):
    x = np.random.uniform(-L/2,L/2,n)
    y = np.random.uniform(-L/2,L/2,n)
    z = np.random.uniform(0,h,n)
    npdentro = np.sum((x**2 + y**2) < (L/2)**2)
    return npdentro
```

```python
L = h = 2

vCilindro = np.pi * (L/2)**2 * h
vPrisma = L**2 * h
lPuntos = [100,1000,10000,100000,1000000]

for n in lPuntos:
    nPuntosDentro = volCilindroMC(L,h,n)
    volCalculadoMC = vPrisma * (nPuntosDentro / n)
    error = errorMC(vCilindro,volCalculadoMC)
    print(f"Con {n} puntos; Volumen: {vCilindro}, Volumen Calculado: {volCalculadoMC}, Error: {error}")
```

## Para Calcular Volumen Cono

```python
def volConoMC(L,h,n):
    x = np.random.uniform(-L/2,L/2,n)
    y = np.random.uniform(-L/2,L/2,n)
    z = np.random.uniform(0,h,n)
    npdentro = np.sum(np.sqrt(x**2 + y**2) < (1 - (z/2)))
    return npdentro
```

```python
L = h = 2

vCono = (1/3) * (np.pi * (L/2)**2 * h)
vPrisma = L**2 * h
lPuntos = [100,1000,10000,100000,1000000]

for n in lPuntos:
    nPuntosDentroCono = volConoMC(L,h,n)
    volCalculadoConoMC = vPrisma * (nPuntosDentroCono / n)
    error = errorMC(vCono,volCalculadoConoMC)
    print(f"Con {n} puntos; Volumen: {vCono}, Volumen Calculado: {volCalculadoConoMC}, Error: {error}")
```

## Para un Solido (Manga Pastelera)

```python
def volSolidoMC(f,L,h,n):
    x = np.random.uniform(-L/2,L/2,n)
    y = np.random.uniform(-L/2,L/2,n)
    z = np.random.uniform(0,h,n)
    npdentro = np.sum(f(x,y,z))
    return npdentro
```

```python
L = h = 2

vSolido = (1/3)*(np.pi * (L/2)**2 * h)
vPrisma = L**2 * h
lPuntos = [100,1000,10000,100000,1000000]

f = lambda x,y,z: z < np.exp(np.sqrt(x**2 + y**2))

for n in lPuntos:
    nPuntosDentroSolido = volSolidoMC(f,L,h,n)
    volCalculadoSolidoMC = vPrisma * (nPuntosDentroSolido / n)
    print(f"Con {n} puntos, Volumen Calculado: {volCalculadoSolidoMC}")
```
