```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.interpolate as sint
import math
%matplotlib inline
```

```python
#Vamos a usar como ejemplo la funcion sin(x) + 1/3 cos(2x)

xf = np.linspace(0,9,1000) #Vector de 0 a 9 con 1000 puntos
yf = np.sin(xf) + ((1/3)*np.cos(2*xf))

plt.figure(figsize = (4,3))
plt.plot(xf,yf)
plt.show()
```

```python
#Vamos a usar como ejemplo la funcion sin(x) + 1/3 cos(2x)

x = np.linspace(0,9,14) #Vector de 0 a 9 con 1000 puntos
y = np.sin(x) + ((1/3)*np.cos(2*x))

plt.figure(figsize = (4,3))
plt.plot(x,y,".")
plt.show()
```

```python
xinterpoladas = np.linspace(0,9,50)
yinterpoladas = np.interp(xinterpoladas,x,y)

plt.figure(figsize = (4,3))
plt.plot(x,y,"+")
plt.plot(xinterpoladas,yinterpoladas,".")
plt.show()
```

```python
xinterpoladas = np.linspace(0,9,50)
fcubica = sint.interp1d(x,y,kind="cubic")
y3interpoladas = fcubica(xinterpoladas)

plt.figure(figsize = (4,3))
plt.plot(x,y,"+")
plt.plot(xinterpoladas,y3interpoladas,".",color="green")
plt.show()
```

```python
plt.plot(xf,yf, label = "sin(x) + (1/3)cos(2x)",color = "blue")
plt.plot(x,y,"+",color = "purple",label = "11 puntos muestrados")
plt.plot(xinterpoladas,yinterpoladas,"*",color = "orange",label = "lineal")
plt.plot(xinterpoladas,y3interpoladas,".",color="green", label = "cubica")
plt.legend()
plt.show()
```

```python
s = sint.InterpolatedUnivariateSpline(x,y,k=4)
yspline= s(xinterpoladas)

plt.plot(xf,yf,color="blue")
plt.plot(xinterpoladas,yspline,color="red")

plt.show()
```
