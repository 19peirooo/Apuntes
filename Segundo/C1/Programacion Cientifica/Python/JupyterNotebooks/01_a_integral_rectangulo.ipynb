# 1a. Integral por el método del rectángulo

Reproducción en python del método más sencillo para integrar numéricamente una función de una variable, que se estudió en la lección introductoria con C.

## Descripción del método

Este método se basa en dividir el rango de integración en intervalos de igual anchura y suponer que la función es constante en ese intervalo, con el valor que toma en el punto inicial.

### Código

```python
# Paquetes que se van a usar
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
num_intervalos = [10,20,50,100,500,1000,10000]
liminf = 0
limsup = 3
valorintegral = (limsup**3-liminf**3)/3
ni = []
errores = []
for nint in num_intervalos:
    h = (limsup-liminf)/nint # Anchura del intervalo
    x = np.linspace(liminf,limsup,nint+1)
    y = x**2
    integralnumerica = np.sum(y[0:-1]*h)
    error = 100*(valorintegral - integralnumerica)/valorintegral
    ni.append(nint)
    errores.append(error)
plt.plot(ni,errores,marker='*')
plt.xscale('log')
plt.show()
```

```python
# Método del rectangulo pasando la función como primer parámetro
def integralrect(f,a,b,n=10):
    h = (b-a)/n # Anchura del intervalo
    xf = np.linspace(a,b,n+1) # para calcular n trapecios necesitamos n+1 puntos
    yf = f(xf)       # La función se ha pasado como parámetro
    integ = np.sum(yf[0:-1]*h)
    return integ
    
ni = []
errores = []
for nint in num_intervalos:
    integralnumerica = integralrect(lambda x: x**2,liminf,limsup,n=nint)
    error = 100*(valorintegral - integralnumerica)/valorintegral
    ni.append(nint)
    errores.append(error)
plt.plot(ni,errores,marker='*')
plt.xscale('log')
plt.show()
```

```python
# Aprovechamos el código para calcular por el mismo método la integral de y = sin(x)
valorintegral = (-np.cos(limsup)+np.cos(liminf))
print(f"Valor integral: {valorintegral}")
ni = []
errores = []
for nint in num_intervalos:
    integralnumerica = integralrect(lambda x: np.sin(x),liminf,limsup,n=nint)
    error = 100*(valorintegral - integralnumerica)/valorintegral
    ni.append(nint)
    errores.append(error)
plt.plot(ni,errores,marker='*')
plt.xscale('log')
plt.show()
```
