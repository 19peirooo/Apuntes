# Diagonalización con Python

```python
#Importamos las librerías necesarias
import numpy as np
from numpy import linalg as LA   #sublibrería de numpy
```

```python
A=np.array([[2,-2,1],[2,-8,-2],[1,2,2]])   #Definimos una matriz A que vamos a diagonalizar
print(A)
```

Utilizaremos la notación: w para los autovalores (valores propios) y v para los autovectores (vectores propios)
Y vamos a utilizar la función eig(A) que nos proporciona autovalores y autovectores.

```python
w,v=LA.eig(A)
w,v
```

```python
np.around(w,2)   # tenemos los valores propios: -7,3,0
```

```python
np.around(v,2)   # y también la matriz de vectores propios escritos por columnas por tanto el primer autovector es (0.24, 0.94,-0.24)
```

Recuerda que puedes tomar cualquier vector de componentes proporcionales (como vector libre sería el mismo) para obtener una notación más cómoda


```python
#Si queremos obtener el primer autovalor
l1=w[0]
l1
```

```python
#y el primer autovector
v1=v[:,0]
v1
```

```python
#y podemos comprobar que efectivamente v1 es vector propio asociado al valor propio l1=-7
print(np.dot(A,v1))
```

```python
print(l1*v1)
```

```python
#Obien podríamos preguntarle
np.dot(A,v1)==l1*v1
```

¿Qué ocurre? que al introducir tantos decimales en el cálculo, en cuanto falle uno aunque sea del orden de 10^^-15
Existe una función isclose que permite comparar números próximos

```python
np.isclose(np.dot(A,v1),l1*v1)   #Así podemos comprobar la igualdad de A*v1 y l1*v1
```

Sabemos que la matriz de autovectores es la matriz regular P que es semejante a la matriz A, es decir tal que P^-1*A*P=D
Vamos a comprobarlo con Python

```python
#Si realizamos el producto de matrices P^-1*A*P; primero calculamos P^-1 y la llamamos vInv
vInv=np.around(np.linalg.inv(v),1)
print(vInv)
```

```python
R=(vInv.dot(A)).dot(v)
print(np.around(R,0))
```

```python
#O bien podríamos haberlo hecho directamente
print(np.around(np.linalg.inv(v).dot(A.dot(v))))
```
