# Manipulación de matrices

Suma de todos los elementos de una matriz: la función np.sum()

```python
M=np.array([[2,-1,0],[2,5,-3],[-1,1,0]])
np.sum(M)
```

Para hacer la suma por filas o por columnas utilizamos la misma función añadiendo el parámetro axis=1  o axis=0 

```python
np.sum(M, axis=0)   #nos da la suma por columnas
```

```python
np.sum(M,axis=1)   #nos da la suma por filas
```

```python
#y el producto de todos los elementos de una matriz
np.prod(M)
```

Podemos calcular también la media aritmética de todos los valores de la matriz, o la media calculada por filas y también la media por columnas

```python
np.mean(M)
```

```python
np.mean(M, axis=0)
```

```python
np.mean(M, axis=1)
```

# Operaciones con matrices

Traspuesta de una matriz. La función .trasnspose(). El punto delante significa que la función no toma la matriz como argumento

El cálculo de la traza de una matriz se hace con la función .trace()

```python
print(M.transpose())
```

```python
print(M.trace())
```

A la hora de realizar operaciones con matrices, todas ellas deben haber sido creadas con la misma librería.

No se pueden operar matrices creadas como listas con otras creadas con numpy

```python
A=np.array([[1,2,-1],[0,1,5]])
B=np.array([[2,3,-2],[3,-1,2]])
print(A+B)
```

```python
print(3*A-2*B)
```

Para hacer el producto de matrices utilizaremos A.dot(B) que no es igual a B.dot(A)

```python
A=np.array([[1,2,-1],[0,1,5]])
C=np.array([[2,1],[1,-1],[0,4]])
print(A.dot(C))
```

```python
#Como en R si realizas la operación A*C se obtiene el resultado de multiplicar elemento a elemento
print(A*B)
```

Potencia de una matriz. Se utiliza la sublibrería linalg y la función matrix_power

```python
P=np.array([[2,-1,3],[4,-1,0],[5,2,-2]])
print(P)
```

```python
print(np.linalg.matrix_power(P,3))   #Calculando P^3
```

# Rango de una matriz

El rango de una matriz es el número de filas no completamente nulas una vez triangularizada la matriz.

El rango es el número de filas (y de columnas) que son linealmente independientes

```python
print(np.linalg.matrix_rank(P))  #Es una matriz 3x3 con rango 3 por tanto rango completo: es una matriz regular
```

# Inversa de una matriz

```python
print(np.linalg.inv(P))
```

```python
#Comprobamos que efectivamente es la matriz inversa de P
Id=(np.linalg.inv(P).dot(P))
print(Id)
```
