# Sistemas de ecuaciones lineales

# 1.-  Sistemas compatibles determinados

## Ejemplo 1
x+y+2z=9

2x+4y-3z=1

3x+6y-5z=0

```python
import numpy as np   #importamos la librería numpy
A=np.array([[1,1,2],[2,4,-3],[3,6,-5]])   #Definimos la matriz de coeficientes
b=np.array([9,1,0])   #Definimos los términos independientes
AB=np.array([[1,1,2,9],[2,4,-3,1],[3,6,-5,0]])
```

Comprobamos que los rangos de matriz de coeficientes y de matriz ampliada son iguales

```python
np.linalg.matrix_rank(A)==np.linalg.matrix_rank(AB)
```

Y comprobamos que ese rango coincide además con el número de incógnitas

```python
np.linalg.matrix_rank(A)==3
```

Aplicando el teorema de Rouché-Fröbenius sabemos por tanto que el sistema es compatible determinado (SCD): tiene solución única

```python
#Resolvemos el sistema
x=np.linalg.solve(A,b)
x
```

Podemos comprobar el resultado: multiplicando la matriz de coeficientes por el vector de soluciones y se obtiene el vector b

```python
A.dot(x)
```

# Ejemplo 1
x+y+2z=9

2x+4y-3z=1

3x+6y-5z=0

```python
from mpl_toolkits.mplot3d import Axes3D 

point1=np.array([0,0,9/2])   #indicamos un punto cualquiera que verifica la ecuación del plano 
n1=np.array([1,1,2])   #e indicamos el vector normal del plano

point2=np.array([0,0,0])
n2=np.array([2,4,-3])

point3=np.array([0,0,-1/3])
n3=np.array([3,6,-5])

# y por último le indicamos los valores de los términos independientes
D1=-9
D2=-1
D3=0
```

```python
X,Y=np.meshgrid(range(30), range(30))   
#En este caso estamos generando toda una malla de coordenadas para x,y y para cada uno de ellos vamos a calcular el valor de z
```

```python
z1=(-n1[0]*X-n1[1]*Y-D1)*1./n1[2]
z2=(-n2[0]*X-n1[1]*Y-D2)*1./n2[2]
z3=(-n3[0]*X-n3[1]*Y-D3)*1./n3[2]  
#Calculamos la z correspondiente a acada punto del plano, de la malla que hemos creado
#multiplicamos por 1. para obtener el valor exacto con decimales y evitar que lo convierta en entero
```

Por último creamos el gráfico 3D correspondiente a nuestro sistema de ecuaciones lineales

```python
plot3d=plt.figure().gca(projection="3d")
plot3d.plot_surface(X,Y,z1, color="blue")   #representamos la primera malla (con puntos tomados del primer plano)
plot3d.plot_surface(X,Y,z2, color="red")
plot3d.plot_surface(X,Y,z3, color="green")
plt.show()
```

Comprobamos que efectivamente son tres planos que se cortan en un único punto

# Ejemplo 4
x+y-z=2

x-y+z=1

3x+y-z=5

```python
A=np.array([[1,1,-1],[1,-1,1],[3,1,-1]])   #Definimos la matriz de coeficientes
b=np.array([2,1,5])   #Definimos los términos independientes
AB=np.array([[1,1,-1,2],[1,-1,1,1],[3,1,-1,5]])
```

```python
np.linalg.matrix_rank(A)==np.linalg.matrix_rank(AB)   #Comprobamos que el sistema es compatible
```

```python
np.linalg.matrix_rank(A)   #Pero el rango es 2 (menor que el número de incógnitas que es 3)
```

Es por tanto un sistema compatible indeterminado de orden 1: tiene infinitas soluciones dependientes de un parámetro, es decir geométricamente son infinitos puntos en una dirección, una recta

```python
from mpl_toolkits.mplot3d import Axes3D 

point1=np.array([0,0,-2])   #indicamos un punto cualquiera que verifica la ecuación del plano 
n1=np.array([1,1,-1])   #e indicamos el vector normal del plano

point2=np.array([0,0,1])
n2=np.array([1,-1,1])

point3=np.array([0,0,-5])
n3=np.array([3,1,-1])

# y por último le indicamos los valores de los términos independientes
D1=-2
D2=-1
D3=-5
```

```python
#Creamos la malla de puntos
X,Y=np.meshgrid(range(-30,30), range(30)) 
X,Y
```

```python
z1=(-n1[0]*X-n1[1]*Y-D1)*1./n1[2]
z2=(-n2[0]*X-n1[1]*Y-D2)*1./n2[2]
z3=(-n3[0]*X-n3[1]*Y-D3)*1./n3[2]  
#Calculamos la z correspondiente a acada punto del plano, de la malla que hemos creado
```

```python
plot3d=plt.figure().gca(projection="3d")
plot3d.plot_surface(X,Y,z1, color="yellow")   #representamos la primera malla (con puntos tomados del primer plano)
plot3d.plot_surface(X,Y,z2, color="red")
plot3d.plot_surface(X,Y,z3, color="cyan")
plt.show()
```

Se trata de tres planos que se cortan en una recta: un haz de planos

# 3.- Sistemas incompatibles

# Ejemplo 5
x+y=2

x-y=1

2x+y=3

```python
A=np.array([[1,1],[1,-1],[2,1]])   
b=np.array([2,1,3])   #Definimos los términos independientes
AB=np.array([[1,1,2],[1,-1,1],[2,1,3]]) 
```

```python
np.linalg.matrix_rank(A)==np.linalg.matrix_rank(AB)
```

Es por tanto un sistema incompatible (teorema de Rouché Fröbenius)

```python
x=np.linspace(-10,10,100)  #permite generar 100 valores entre -10 y 10
plt.plot(x, 2-x, x, x-1, x, 3-2*x)
```
